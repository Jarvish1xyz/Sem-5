# ====> Bank Loan Default Prediction Using Machine Learning

#### Banks provide loans to customers, but some customers fail to repay them, causing financial losses. Predicting whether a customer is likely to default before approving a loan can help banks reduce risk and make better lending decisions.

#### Objective:
#### Develop a machine learning model that predicts whether a customer will default on a loan based on their personal, financial, and loan-related information.
#### ===========================================================================================

#### => Dataset Summary
#### Dataset Name: Loan Default Prediction Dataset

#### Domain: Banking / Finance

#### Target Variable: Default (or Loan_Status, depending on the dataset)

#### 0 = No Default
#### 1 = Default

=> Feature 
| Feature        | Description                                       | Type                 |
| -------------- | ------------------------------------------------- | -------------------- |
| LoanID         | Unique loan ID                                    | Categorical          |
| Age            | Age of the applicant                              | Numerical            |
| Income         | Annual income of the applicant                    | Numerical            |
| LoanAmount     | Amount of loan requested                          | Numerical            |
| CreditScore    | Applicant's credit score                          | Numerical            |
| MonthsEmployed | Total months employed                             | Numerical            |
| NumCreditLines | Number of active credit lines                     | Numerical            |
| InterestRate   | Loan interest rate (%)                            | Numerical            |

| DTIRatio       | Debt-to-Income ratio                              | Numerical            |
| Education      | Highest education level                           | Categorical          |
| EmploymentType | Type of employment                                | Categorical          |
| MaritalStatus  | Marital status                                    | Categorical          |
| HasMortgage    | Whether the applicant has a mortgage              | Categorical (Yes/No) |
| HasDependents  | Whether the applicant has dependents              | Categorical (Yes/No) |
| LoanPurpose    | Purpose of the loan                               | Categorical          |
| HasCoSigner    | Whether the loan has a co-signer                  | Categorical (Yes/No) |
| **Default**    | **Target variable (0 = No Default, 1 = Default)** | **Target**           |
