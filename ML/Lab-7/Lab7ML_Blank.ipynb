# Decision Tree Classifier

```python
import numpy as np
from sklearn.datasets import load_iris
```

# Import iris data ser using sklearn

```python
iris = load_iris()
display(iris)
```

# Importing the libraries

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

# Divide the data into input and output

```python
x=iris.data
y=iris.target
```

```python
display(x)
display(y)
```

# Splitting the dataset into the Training set and Test set

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

# Fitting DecisionTreeClassifier   on dataset

```python
dtree = DecisionTreeClassifier()
dtree.fit(X_train, y_train)
```

# Display Decision Tree

```python
plt.figure(figsize=(15,20))
plot_tree(dtree, filled=True, feature_names=iris.feature_names, class_names=iris.target_names)
plt.show()
```

# Predict the x_test 

```python
y_predict = dtree.predict(X_test)
display(y_predict)
```

```python
display(y_test)
```

# Display Training Accuracy

```python
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix

train_accuracy = accuracy_score(y_train, dtree.predict(X_train))
print(f"Training Accuracy: {train_accuracy:.2f}")
```

# Display Test Accuracy

```python
test_accuracy = accuracy_score(y_test, y_predict)
print(f"Testing Accuracy: {test_accuracy:.2f}")
```

# Confusion Matrix

```python
cm = confusion_matrix(y_test, y_predict)
print("Confusion Matrix:")
display(cm)
```

```python
cr = classification_report(y_test, y_predict)
print("Classification Report:")
print(cr)
```

# Save Model State

```python
import pickle
fileName = "dtree_model.pkl"
pickle.dump(dtree, open(fileName, 'wb'))
print(f"Model saved to {fileName}")
```

# Load Model from Disk

```python
loaded_model = pickle.load(open(fileName, 'rb'))
print("Model loaded from successfully:")
```

```python
loaded_model_predict = loaded_model.predict(X_test)
print("Predictions from loaded model:")
display(loaded_model_predict)
```
