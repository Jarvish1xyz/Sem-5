# Importing the libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

# Importing the libraries

```python
df = pd.read_csv("Position_Salaries.csv")
```

```python
df
```

```python
df[['Level','Salary']]
```

# Plot the data using scatter plot

```python
plt.scatter(df['Level'],df['Salary'])
```

```python
x = df.drop(columns=['Salary', 'Position'])
x
y = df.Salary
y
```

# Splitting the dataset into the Training set and Test set

```python
xTrain, xTest, yTrain, yTest = train_test_split(x, y, test_size=0.3, random_state=42)
```

```python
xTest
```

# Fitting Linear Regression to the dataset

```python
model = LinearRegression()
```

```python
model.fit(xTrain, yTrain)
```

# Predict the x_test using Linear Model

```python
yPredict = model.predict(xTest)
yPredict
```

# Visualising the Linear Regression results

```python
yPredict
```

```python
plt.scatter(x,y, color="red", label="Actual Data")
plt.plot(xTest, yPredict, label="Predict Data")
plt.legend()
```

```python
plt.scatter(xTest, yTest, color="Red", label="Actual Data")
plt.plot(xTest, yPredict, label="Predict Data")
plt.legend()
```

# Fitting Polynomial Regression to the dataset

## import  PolynomialFeatures

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

## create PolynomialFeatures of degree 4

```python
df = pd.read_csv("Position_Salaries.csv")
```

```python
x = df.drop(columns=['Salary', 'Position'])
x
y = df.Salary
y
```

```python
poly = PolynomialFeatures(degree=4)
```

```python
xTrain, xTest, yTrain, yTest = train_test_split(x, y, test_size=0.3, random_state=42)
```

## Apply fit_transform to features

```python
xTrainPoly = poly.fit_transform(xTrain)
xTestPoly = poly.transform(xTest)
```

## View new  Features

```python
poly.fit_transform(x)
```

## Fitting Polynomial Regression to the dataset

```python
polyModel = LinearRegression()
```

```python
polyModel.fit(xTrainPoly, yTrain)
```

# Create Grid for higher resolution and smoother curve

```python
import numpy as np
x1 = np.arange(1, 10, 0.1)
x1 = x1.reshape(-1,1)
x1
```

```python
plt.scatter(x,y,color='red', label="Actual Salaries")
plt.plot(x1, polyModel.predict(poly.transform(x1)), color='blue', label="Polynomial Regression")
plt.title("Polynomial Regression (Level vs Salary)")
plt.xlabel('Level')
plt.ylabel('Salary')
plt.legend()
plt.show()
```

# Predict the x_test using Polynomial Regression

```python
yPredictPoly = polyModel.predict(xTestPoly)
yPredictPoly
```
